## For NMR data

1) Find NMR files from Biological Magnetic Resonance Data Bank (https://bmrb.io/). E.g. vanillin (https://bmrb.io/metabolomics/mol_summary/show_data.php?id=bmse000343&whichTab=1) or Human Metabolome Database (https://www.hmdb.ca/)

3) Download the .zip file under Time Domain Data. Within the file, there should be bmseXXXXXXX/nmr/set01/1H, and the like. For Topspin to recognize the file, need to rename to just a number (e.g. 1H become 1, 13C becomes C).

4) Open the numbered file in Topspin. Process as normal, importantly, input the command "convbin2asc". This is create as ascii-spec.txt file in the same directory. The file should have 4 columns: Point number, Intensity, Chemical Shift (Hz), Chemical Shift (ppm). (https://u-of-o-nmr-facility.blogspot.com/2014/02/getting-xy-ascii-data-from-topspin.html)

6) Rename the .txt file to a .csv file. In excel, truncate to just "Chemical Shift (ppm)" and "Intensity" columns, add header row with those labels.


```python
nmrH = pd.read_csv('5/chloroxylenol_HNMR.csv')
nH = 8.2
nmrH
```

```python
nmrH1 = round(nmrH['Chemical Shift (ppm)'],3)
nmrH['Chemical Shift (ppm)']=nmrH1
nmrH2 = nmrH.groupby('Chemical Shift (ppm)')['Intensity'].mean().reset_index()
nmrH3 = round(nmrH2['Intensity']/nmrH2['Intensity'].max()*100,2)
nmrH2['Intensity'] = nmrH3
nmrH4 = nmrH2[abs(nmrH2['Intensity']) > -0.1]

nmrH5 = nmrH4['Intensity'].cumsum()
nmrH5 = round(nH - nmrH5/nmrH5.max()*nH,2)
nmrH4['Integral'] = nmrH5
nmrH4
```

```python
nmrH4.to_csv('5/5_HNMR.csv',index=False)
```

```python
nmrC = pd.read_csv('5/chloroxylenol_CNMR.csv')
nmrC
```

```python
nmrC1 = round(nmrC['Chemical Shift (ppm)'],3)
nmrC['Chemical Shift (ppm)']=nmrC1
nmrC2 = nmrC.groupby('Chemical Shift (ppm)')['Intensity'].mean().reset_index()
nmrC3 = round(nmrC2['Intensity']/nmrC2['Intensity'].max()*100,2)
nmrC2['Intensity'] = nmrC3
nmrC4 = nmrC2[abs(nmrC2['Intensity']) > -0.1]
nmrC4
```

```python
nmrC4.to_csv('5/5_CNMR.csv',index=False)
```

## For MS Data

1) Find compound in https://sdbs.db.aist.go.jp/
2) Navigate to the MS data. There should be a button "peak data" below the spectrum itself.
3) Copy and paste the MS peaks into an excel file. Add header rows that specify "m/z" and "Intensity"
4) Save excel file as .csv file

```python
ms = pd.read_csv('5/chloroxylenol_MS.csv')
print(ms)
```

```python
ms1=[]
ms2=[]
for i in range(ms['m/z'].min()-5,ms['m/z'].max()+10):
    ms1.append(i)
    if i in ms['m/z'].to_list():
        ms2.append(ms['Intensity'].loc[list(ms['m/z']).index(i)])
    else:
        ms2.append(0)

ms3=pd.DataFrame({'m/z':ms1,'Intensity':ms2})
ms3
```

```python
ms3.to_csv('5/5_MS.csv',index=False)
```
